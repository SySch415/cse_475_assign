## Load dataset

Now, we are importing a dataset of diabetes. You can check the details on this dataset here: https://scikit-learn.org/stable/datasets/toy_dataset.html#diabetes-dataset.

The dataset consists of 442 observations with 10 attributes ($X$) that may affect the progression of diabetes ($y$). Ten baseline variables, age, sex, body mass index, average blood pressure, and six blood serum measurements were obtained for each of $n$ = 442 diabetes patients, as well as the response of interest, a quantitative measure of disease progression one year after baseline.

````python
# Load the diabetes dataset
diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True)
print('The shape of the input features:',diabetes_X.shape)
print('The shape of the output varaible:',diabetes_y.shape)
````

We will choose just one attribute from the ten attributes as an input variable.

````python
# Use only one feature
diabetes_X_one = diabetes_X[:, np.newaxis, 2]
print(diabetes_X_one.shape)
````

## Dataset split

Now, we split the dataset into two parts: training set and test set.

- training set: 422 samples
- test set: 20 samples

````python
# Split the data into training/testing sets
diabetes_X_train = diabetes_X_one[:-20]
diabetes_X_test = diabetes_X_one[-20:]

# Split the targets into training/testing sets
diabetes_y_train = diabetes_y[:-20]
diabetes_y_test = diabetes_y[-20:]

print('Training input variable shape:', diabetes_X_train.shape)
print('Test input variable shape:', diabetes_X_test.shape)
````

## Linear regression

Assume that we have a hypothesis $$ h_{\theta}(x) = \theta_0 + \theta_1 x. $$

Your tasks:

Implementation tasks:

- [Coding][4pts] implement "your own version" of the method of least-squares, compute and report $\theta_0$ and $\theta_1$ that minimize the residual sum of squares,
$$ \sum_{i=1}^{N} \frac{1}{2}( y^{(i)} - h_{\theta}(x^{(i)}))^2$$

[IMPORTANT] Do not just call the least square function from libraries, for example,
scipy.optimize.least_squares from scipy. Doing so will result in 0 point. Using helping functions such as numpy.linalg.inv is okay.

**Hint for least squares.** Write the prediction model in matrix form as $\hat{\mathbf{y}} = X\theta$. For the training set, construct
$$
X = \begin{bmatrix}
1 & x^{(1)} \\
1 & x^{(2)} \\
\vdots & \vdots \\
1 & x^{(n_{\mathrm{train}})}
\end{bmatrix}, \qquad
\mathbf{y} = \begin{bmatrix}
y^{(1)} \\
y^{(2)} \\
\vdots \\
y^{(n_{\mathrm{train}})}
\end{bmatrix}, \qquad
\theta = \begin{bmatrix}\theta_0 \\ \theta_1\end{bmatrix}.
$$
Here, $X$ is the design matrix formed from `diabetes_X_train` with a column of ones added for the intercept, and $\mathbf{y}$ is `diabetes_y_train` written as a column vector. Thus, $X$ has shape $(n_{\mathrm{train}},2)$ and $\mathbf{y}$ has shape $(n_{\mathrm{train}},1)$, where $n_{\mathrm{train}}=422$. Compute $\theta$ by solving the normal equations
$$
X^{\intercal}X\theta = X^{\intercal}\mathbf{y}.
$$
Construct these quantities yourself and solve this system of equations; do not use a library least-squares estimator.

- [Coding][6pts] implement your own version of the gradient descent algorithm, compute and report $\theta_0$ and $\theta_1$ that minimize the mean squared error $$ \sum_{i=1}^{N} \frac{1}{N}( y^{(i)} - h_{\theta}(x^{(i)}))^2$$

[NOTE] Notice that the loss function is mean-squared error. Implement the gradient descent update rule in a for loop.

**Hint for gradient descent.** Initialize $\theta$ randomly, choose a learning rate $\eta$ and a number of iterations, and use the following structure:

```text
initialize theta randomly
for t = 1, ..., number_of_iterations:
    compute the analytical gradient grad_theta_L using all training samples
    theta <- theta - eta * grad_theta_L
```

The gradient must be derived analytically from the stated MSE loss. Since $\theta=[\theta_0,\theta_1]^{\intercal}$, organize it as
$$
\nabla_{\theta}L(\theta)
= \frac{1}{n_{\mathrm{train}}}\sum_{i=1}^{n_{\mathrm{train}}}
\begin{bmatrix}
\text{element for }\theta_0 \\
\text{element for }\theta_1
\end{bmatrix}.
$$
Determine the two elements from the hypothesis and loss function. Refer to the lecture slides for the required partial derivatives.

To check whether your computations (LS/GD) are correct, consider using an API such as Scikit learn linearregression to see if you get similar results --- the numbers may not be exactly same, which is okay as far as they are "close" enough (for example, matching first two significant digits).
